```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("housing.csv")
```

```python
data = pd.read_csv("housing.csv")
data["income_cat"] = pd.cut(data["median_income"],
                            bins=[0, 1.5, 3.0, 4.5, 6.0, np.inf],
                            labels=[1, 2, 3, 4, 5])
```

```python
from sklearn.model_selection import StratifiedShuffleSplit

split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in split.split(data,data["income_cat"]):
    strat_train_set = data.loc[train_index]
    strat_test_set = data.loc[test_index]
    
```

```python
# Removing the Income Cat column
for sett in (strat_train_set,strat_test_set):
    sett.drop('income_cat',axis=1,inplace=True)
```

```python
df = strat_train_set.copy()
```

### Handling Missing Data
Some features, like total_bedrooms , contain missing values. You can:
- Drop rows with missing values
- Drop the entire column
- Impute missing values (recommended)
We’ll use option 3 using SimpleImputer from Scikit-Learn, which allows consistent
handling across all datasets (train, test, new data):

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="median")

hosuing_num = housing.select_dtypes(include=[np.number]) # Here we have removed the ocean_proximity column
X = imputer.fit_transform(hosuing_num)
```
