```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("../Data/housing.csv")
```

```python
data["income_cat"] = pd.cut(data["median_income"],
                            bins=[0, 1.5, 3.0, 4.5, 6.0, np.inf],
                            labels=[1, 2, 3, 4, 5])
```

```python
from sklearn.model_selection import StratifiedShuffleSplit

split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in split.split(data,data["income_cat"]):
    strat_train_set = data.loc[train_index]
    strat_test_set = data.loc[test_index]
    
```

```python
# Removing the Income Cat column
for sett in (strat_train_set,strat_test_set):
    sett.drop('income_cat',axis=1,inplace=True)
```

```python
df = strat_train_set.copy()
```

```python
# df.info()
# We can see there are some missing values in the bedrooms values
```

```python
# First we will seprate our labels and the data 
housing = strat_train_set.drop("median_house_value",axis=1)
housing_labels = strat_train_set['median_house_value'].copy()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="median")

housing_num = housing.select_dtypes(include=[np.number]) # Here we have removed the ocean_proximity column
X = imputer.fit_transform(housing_num)
```

```python
# Even if you input a Pandas DataFrame, the output of transformers like transform() will be a NumPy array. converting it back like this:

housing = pd.DataFrame(X, columns=housing_num.columns, index=housing_num.index)

# Adding Ocean_Proximity column which we removed earlier
housing['ocean_proximity'] = df['ocean_proximity']
```

```python
housing = housing[['ocean_proximity']]
# housing
```

##### Text columns like "ocean_proximity" are not free-form text but limited to a fixed set of values (e.g., "NEAR BAY" , "INLAND" ). These are known as categorical attributes.

---

#### For unordered categories, one-hot encoding is a better choice. It creates one binary column per category.

```python
from sklearn.preprocessing import OneHotEncoder

cat_encoder = OneHotEncoder()
housing_cat = cat_encoder.fit_transform(housing)
```

```python
housing_cat.toarray()
```

```python
cat_encoder.categories_
```

```python
housing_cat = pd.DataFrame(housing_cat.toarray(), columns=['<1H OCEAN', 'INLAND', 'ISLAND', 'NEAR BAY', 'NEAR OCEAN'], index=housing.index)
```

```python
housing_cat
```

```python
df = pd.concat([df,housing_cat],axis=1)
```
